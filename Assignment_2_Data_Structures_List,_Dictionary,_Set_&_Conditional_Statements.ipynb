**LISTS**

**1. List Creation**

```python
age_list = [24, 25, 26, 27, 28]
name_list = ["Zara", "Maya", "Liam", "Noah", "Ava"]
```

**2. List Operations / Modifications**

```python
# a. Append
name_list.append("Yazhini")
```

```python
# b. Insert 30 at index 2
age_list.insert(2, 30)
```

```python
# c. Remove Yazhini
name_list.remove("Yazhini")
```

```python
# d. Pop last element
age_list.pop()
```

```python
# e. Extend with more ages
age_list.extend([29, 30, 26])
```

```python
# f. Sort descending
age_list.sort(reverse=True)
```

```python
# g. Max, Min, Sum
print("Max age:", max(age_list))
print("Min age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

**3. Accessing List Elements**

```python
print("First element:", name_list[0])
print("Last element:", name_list[-1])
print("Index 2 to 4:", name_list[2:5])
print("Reverse order:", name_list[::-1])
```

**DICTIONARIES**

```python
# a. Create dictionary
student_marks = {
    "Ravi": 85,
    "Keerthi": 92,
    "Vishal": 78,
    "Divya": 88,
    "Renu": 90
}
```

```python
# b. Print specific student's mark
print("Mark of Keerthi:", student_marks["Keerthi"])
```

```python
# c. Add Janani
student_marks["Janani"] = 80
```

```python
# d. Update one student’s mark
student_marks["Vishal"] = 82
```

```python
# e. Print keys, values, items
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
```

**Sets (Operations):**


```python
# a. Create a set

my_set = {'a','e','i','o','u','a','a','i'}
print(my_set)
```

**Explanation:**

Sets store only unique values.
Duplicate 'a' and 'i' are removed automatically, so the output will be:
{'a', 'e', 'i', 'o', 'u'} .

```python
# b. Attempt to modify

my_set[4] = 's'
```

**Explanation:**

Sets do not support indexing,Cannot access or modify elements by index.
They are unordered and immutable in terms of positions.

```python
# c. Create two sets

set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
```

```python
# d. Union & Intersection

print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

**OPERATORS & CONDITIONAL STATEMENTS**

```python
score = int(input("Enter your score (0 to 10): "))

if score > 7:
    print("Above Average: Excellent! Keep up the great performance.")
elif 4 <= score <= 7:
    print("Average: Good effort! Keep practicing, there's room for improvement.")
elif score < 4:
    print("Below Average: Need to improve your performance. Consistent practice helps.")
else:
    print("Invalid score entered.")
```
