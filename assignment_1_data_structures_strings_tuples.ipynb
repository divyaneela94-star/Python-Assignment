**1. Strings — Concatenation**

```python
# String Concatenation
str1 = "Hello "
name = input("Enter your Name: ")
result = str1 + name
print(result)
```

```python
# Concatenate additional string
str3 = ", welcome to Python programming"
result2 = result + str3
print(result2)
```

**2. String Slicing and Indexing**

Assuming result2 from above is used:

```python
text = result2

# a. First character
print("First character:", text[0])
```

```python
# b. Last character
print("Last character:", text[-1])
```

```python
# c. First 5 characters
print("First 5 characters:", text[:5])
```

```python
# d. Last 11 characters
print("Last 11 characters:", text[-11:])
```

```python
# e. Reverse the string
print("Reversed string:", text[::-1])
```

```python
# f. Slice the word “Python”
print("Extracted word:", text[text.find("Python"):text.find("Python") + 6])
```

**3. String Methods**

```python
strM = "Python beginner tutorial"
```

```python
# a. Uppercase
print(strM.upper())
```

```python
# b. Lowercase
print(strM.lower())
```

```python
# c. Capitalize
print(strM.capitalize())
```

```python
# d. Count occurrences of 't'
print("Occurrences of 't':", strM.count('t'))
```

```python
# e. Replace "Python" with "Machine Learning"
print(strM.replace("Python", "Machine Learning"))
```

**4. Tuples — Creation, Modification, Access**

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)
```

```python
# a. Concatenate
t_combine = t1 + t2
print("Combined tuple:", t_combine)
```

```python
# b. Repeat 3 times
print("Repeated tuple:", t_combine * 3)
```

```python
# c. 3rd element (index 2)
print("3rd element:", t_combine[2])
```

```python
# d. First 3 elements
print("First 3 elements:", t_combine[:3])
```

```python
# e. Last 3 elements
print("Last 3 elements:", t_combine[-3:])
```
